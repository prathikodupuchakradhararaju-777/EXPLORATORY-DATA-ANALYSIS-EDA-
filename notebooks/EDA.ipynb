# Step 1: Key Performance Indicators (KPIs)

## What are KPIs?

Key Performance Indicators (KPIs) are numerical values that provide a quick summary of the dataset.

In this dashboard, the following KPIs will be calculated:

- Total Machines
- Machine Failures
- Average Air Temperature
- Average Torque

These KPIs help users understand the dataset at a glance.

```python
# Calculate KPI values

total_machines = len(df)

machine_failures = df["Machine failure"].sum()

average_air_temperature = df["Air temperature [K]"].mean()

average_torque = df["Torque [Nm]"].mean()
```

```python
print("Total Machines :", total_machines)

print("Machine Failures :", machine_failures)

print("Average Air Temperature :", round(average_air_temperature, 2))

print("Average Torque :", round(average_torque, 2))
```

# Step 3: Machine Type Distribution

## Objective

This visualization displays the distribution of different machine types available in the dataset.

The chart helps analyze the number of machines belonging to each category (L, M, and H).

An interactive Plotly bar chart is used for visualization.


```python
# Machine Type Distribution

type_chart = px.histogram(
    data_frame=df,
    x="Type",
    color="Type",
    title="Machine Type Distribution",
    text_auto=True
)

type_chart.update_layout(
    xaxis_title="Machine Type",
    yaxis_title="Number of Machines",
    template="plotly_white"
)

type_chart.show()
```

# Step 4: RPM and Tool Wear Analysis

## Objective

The objective of this step is to analyze the relationship between the rotational speed (RPM) and tool wear of the machines.

An interactive scatter plot is used to identify patterns and determine whether tool wear increases or decreases with changes in rotational speed.

This analysis helps in understanding machine performance and maintenance behavior.

```python
# Step 4: RPM vs Tool Wear Analysis

rpm_toolwear_chart = px.scatter(
    data_frame=df,
    x="Rotational speed [rpm]",
    y="Tool wear [min]",
    color="Machine failure",
    title="RPM vs Tool Wear Analysis",
    labels={
        "Rotational speed [rpm]": "Rotational Speed (RPM)",
        "Tool wear [min]": "Tool Wear (Minutes)"
    }
)

rpm_toolwear_chart.update_layout(
    template="plotly_white"
)

rpm_toolwear_chart.show()
```

## Observation

- The scatter plot visualizes the relationship between rotational speed and tool wear.
- Each point represents an individual machine observation.
- Different colors indicate whether the machine experienced a failure.
- This visualization helps identify trends, clusters, and possible relationships between RPM, tool wear, and machine failure.

# Step 5: Sidebar Filters

## Objective

Sidebar filters allow users to interactively filter the dataset based on different machine characteristics.

The dashboard will support filtering by:

- Machine Type
- Machine Failure Status

These filters help users analyze specific subsets of the dataset without modifying the original data.

```python
# Filter data by Machine Type

machine_type = "L"   # Change to M or H

filtered_df = df[df["Type"] == machine_type]

print("Selected Machine Type:", machine_type)
print("Number of Records:", len(filtered_df))

filtered_df.head()
```

```python
# Filter data by Machine Failure

failure_status = 1    # 1 = Failure, 0 = No Failure

filtered_failure = df[df["Machine failure"] == failure_status]

print("Failure Status:", failure_status)
print("Number of Records:", len(filtered_failure))

filtered_failure.head()
```

## Observation

- Filtering allows users to analyze only the required subset of the dataset.
- Machine Type filtering displays records for a selected machine category.
- Machine Failure filtering displays either failed or non-failed machines.
- Interactive dashboards use sidebar filters to perform these operations dynamically.
